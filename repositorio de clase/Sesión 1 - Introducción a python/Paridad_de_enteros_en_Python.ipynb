# 1. Introducción teórica

## 1.1. El algoritmo de la división

Sean $a$ y $b$ números enteros, con $b\neq 0$. El **algoritmo de la división** afirma que existen enteros únicos $q$ y $r$ tales que

$$
a=bq+r, \qquad 0\leq r<|b|.
$$

Los números $q$ y $r$ reciben los nombres de **cociente** y **residuo**, respectivamente. Si dividimos entre $2$, la condición $0\leq r<2$ implica que el residuo solo puede ser $0$ o $1$. Esta observación es la base de la clasificación de los enteros según su paridad.

Por ejemplo:

$$
14=2(7)+0, \qquad 9=2(4)+1.
$$

La idea también funciona con enteros negativos:

$$
-6=2(-3)+0, \qquad -7=2(-4)+1.
$$

## 1.2. La relación de divisibilidad

Dados dos enteros $d$ y $a$, con $d\neq 0$, decimos que **$d$ divide a $a$** si existe un entero $k$ tal que

$$
a=dk.
$$

Esta relación se escribe $d\mid a$. Si no existe tal entero, escribimos $d\nmid a$.

Por ejemplo, $3\mid 12$ porque $12=3(4)$, mientras que $3\nmid 14$. En términos del algoritmo de la división, $d\mid a$ exactamente cuando el residuo de dividir $a$ entre $d$ es cero.

## 1.3. Divisibilidad y paridad

Un entero $n$ es **par** cuando $2\mid n$. Equivalentemente, existe $k\in\mathbb{Z}$ tal que

$$
n=2k.
$$

Un entero $n$ es **impar** cuando $2\nmid n$. Por el algoritmo de la división, esto equivale a que existe $k\in\mathbb{Z}$ tal que

$$
n=2k+1.
$$

Por tanto, todo entero pertenece exactamente a una de estas dos categorías. En Python, `n % 2` calcula el residuo de dividir `n` entre `2`:

- si `n % 2 == 0`, entonces $n$ es par;
- si `n % 2 == 1`, entonces $n$ es impar.

Como usaremos divisor positivo (`2`), Python devuelve residuo `0` o `1` incluso si `n` es negativo. Además, el cero es par porque $0=2(0)$.

````python
# Comprobamos el residuo en algunos enteros positivos, negativos y en cero.
for numero in [14, 9, 0, -6, -7]:
    print(f"{numero:>3} % 2 = {numero % 2}")
````

La salida anterior traduce directamente la definición matemática al lenguaje de Python: los residuos iguales a `0` corresponden a números pares y los residuos iguales a `1`, a números impares.

# 2. Planteamiento del ejemplo aplicado

Dada una lista de números enteros, queremos contar cuántos son pares y cuántos son impares. El resultado tendrá la forma de un diccionario:

```python
{"pares": cantidad_de_pares, "impares": cantidad_de_impares}
```

Para resolverlo seguiremos tres pasos conceptuales:

1. visitar cada número de la lista;
2. calcular su residuo al dividirlo entre `2`; y
3. aumentar el contador correspondiente.

Usaremos la misma lista en todos los enfoques. Así podremos comparar las soluciones sin cambiar el problema. La lista contiene enteros positivos, negativos y el número cero.

````python
numeros = [12, 7, 0, -3, 18, 25, -8, 11, 42, 5, -14, 9]
numeros
````

# 3. Solución mediante programación imperativa

La programación imperativa describe explícitamente la secuencia de instrucciones que modifica el estado del programa. En este ejercicio, el estado está representado por los dos contadores guardados en el diccionario.

## 3.1. Enfoque 1: una sola función para todo el proceso

La función `contar_paridad_imperativo` reúne todas las tareas: crea el diccionario, recorre la lista, clasifica cada entero con `if-else`, actualiza los contadores y devuelve el resultado. Es una solución directa y fácil de seguir para un ejercicio pequeño.

````python
def contar_paridad_imperativo(lista_numeros):
    conteo = {"pares": 0, "impares": 0}

    for numero in lista_numeros:
        if numero % 2 == 0:
            conteo["pares"] += 1
        else:
            conteo["impares"] += 1

    return conteo


resultado_enfoque_1 = contar_paridad_imperativo(numeros)
resultado_enfoque_1
````

El ciclo `for` toma un elemento a la vez. Cuando el residuo es cero, la instrucción `conteo["pares"] += 1` suma una unidad al contador de pares; en caso contrario, el bloque `else` incrementa el contador de impares. Al terminar el recorrido, el diccionario resume toda la lista.

## 3.2. Enfoque 2: separar la clasificación y el conteo

Ahora dividimos el problema en dos funciones. `identificar_paridad` se ocupa únicamente de clasificar un entero. `contar_paridad_modular` recorre la lista, reutiliza la primera función y actualiza el diccionario. Cada función tiene una responsabilidad concreta.

````python
def identificar_paridad(numero):
    if numero % 2 == 0:
        return "par"
    else:
        return "impar"


def contar_paridad_modular(lista_numeros):
    conteo = {"pares": 0, "impares": 0}

    for numero in lista_numeros:
        paridad = identificar_paridad(numero)

        if paridad == "par":
            conteo["pares"] += 1
        else:
            conteo["impares"] += 1

    return conteo


resultado_enfoque_2 = contar_paridad_modular(numeros)
resultado_enfoque_2
````

Podemos probar `identificar_paridad` de manera independiente. Esto es útil porque permite revisar la regla central del programa sin ejecutar todo el conteo.

````python
for numero in [4, 7, -2, -5, 0]:
    print(f"{numero:>2} es {identificar_paridad(numero)}")
````

## 3.3. ¿Cuál enfoque es mejor?

Los dos enfoques son correctos y recorren la lista una vez, por lo cual su costo crece linealmente con la cantidad de elementos: $O(n)$. La diferencia principal no está en el resultado ni en la eficiencia, sino en la organización del código.

| Criterio | Una sola función | Dos funciones |
|---|---|---|
| Lectura inicial | Muy directa para un ejemplo corto | Requiere seguir una llamada adicional |
| Responsabilidades | Mezcla clasificación y conteo | Separa clasificación y conteo |
| Reutilización | La regla de paridad queda dentro del conteo | `identificar_paridad` puede usarse en otros ejercicios |
| Pruebas | Se prueba todo el proceso a la vez | Cada parte puede probarse de forma aislada |
| Mantenimiento | Adecuado para código muy pequeño | Más conveniente si el programa crece |

**Conclusión.** Para una demostración mínima, el primer enfoque es suficiente. Sin embargo, el segundo suele ser mejor diseño porque separa responsabilidades, facilita las pruebas y permite reutilizar la clasificación. Esa ventaja se vuelve más importante a medida que el programa crece.

# 4. Solución mediante programación orientada a objetos

En la programación orientada a objetos agrupamos datos y comportamiento. La clase `Enteros` almacenará un número en el atributo `numero`, mientras que el método `paridad` describirá una operación propia de ese objeto. Aunque el nombre de la clase está en plural para seguir el enunciado, cada instancia representa un solo entero.

## 4.1. Clase `Enteros`

El método especial `__init__` se ejecuta al crear una instancia. Primero verifica que el argumento sea un entero y luego lo guarda. El método `paridad` aplica exactamente la misma regla matemática usada en las funciones anteriores. `__repr__` solo proporciona una representación legible de cada objeto.

````python
class Enteros:
    def __init__(self, numero):
        if not isinstance(numero, int):
            raise TypeError("El valor debe ser un número entero.")
        self.numero = numero

    def paridad(self):
        if self.numero % 2 == 0:
            return "par"
        else:
            return "impar"

    def __repr__(self):
        return f"Enteros({self.numero})"


ejemplo_entero = Enteros(7)
print(ejemplo_entero)
print(ejemplo_entero.paridad())
````

La instancia `Enteros(7)` conserva el valor `7` y puede responder por su propia paridad mediante `ejemplo_entero.paridad()`. Si intentamos construirla con un valor que no sea de tipo `int`, la clase genera un error claro.

## 4.2. Transformar una lista de números en objetos

La siguiente función recorre la lista original. En cada iteración construye un objeto `Enteros` y lo agrega a una nueva lista. La lista de entrada no se modifica.

````python
def transformar_en_enteros(lista_numeros):
    lista_objetos = []

    for numero in lista_numeros:
        objeto_entero = Enteros(numero)
        lista_objetos.append(objeto_entero)

    return lista_objetos


objetos_enteros = transformar_en_enteros(numeros)
objetos_enteros
````

El resultado ya no es una lista de valores de tipo `int`, sino una lista de instancias de `Enteros`. Cada una contiene un número y dispone del método `paridad`.

## 4.3. Contar la paridad de los objetos

La función de conteo recibe ahora objetos en lugar de enteros básicos. Por eso no calcula directamente `numero % 2`: delega la clasificación al método `objeto.paridad()`.

````python
def contar_paridad_objetos(lista_objetos):
    conteo = {"pares": 0, "impares": 0}

    for objeto in lista_objetos:
        if objeto.paridad() == "par":
            conteo["pares"] += 1
        else:
            conteo["impares"] += 1

    return conteo


resultado_objetos = contar_paridad_objetos(objetos_enteros)
resultado_objetos
````

Esta solución distribuye el trabajo entre la clase y las funciones: el objeto conoce su paridad, la primera función construye los objetos y la segunda los cuenta. Para un problema tan pequeño, este diseño tiene más piezas que la solución imperativa; su valor pedagógico consiste en mostrar cómo los datos y su comportamiento pueden reunirse en una clase.

# 5. Combinación de programación funcional y orientación a objetos (opcional)

Python permite combinar estilos. En la siguiente solución, una sola función recibe la lista de enteros, crea temporalmente los objetos, obtiene sus paridades y construye el diccionario final.

Usaremos dos herramientas:

- `map` aplica una transformación a cada elemento. Aquí la función `lambda` crea un objeto `Enteros` y llama inmediatamente a su método `paridad`;
- `Counter`, de la biblioteca estándar `collections`, cuenta cuántas veces aparece cada resultado.

No necesitamos escribir una función separada para construir la lista de objetos, ni conservar esa lista en memoria.

````python
from collections import Counter


def contar_paridad_funcional(lista_numeros):
    paridades = map(lambda numero: Enteros(numero).paridad(), lista_numeros)
    conteo = Counter(paridades)

    return {"pares": conteo["par"], "impares": conteo["impar"]}


resultado_funcional = contar_paridad_funcional(numeros)
resultado_funcional
````

`map` produce las etiquetas `"par"` e `"impar"` de manera perezosa: genera cada resultado cuando `Counter` lo solicita. Después, `Counter` cuenta las etiquetas y la función adapta sus claves al formato pedido: `"pares"` e `"impares"`. Si una categoría no aparece, `Counter` devuelve cero para ella, por lo que la función también funciona con listas que contienen únicamente pares, únicamente impares o ningún elemento.

# 6. Verificación y cierre

Finalmente, verificamos que todos los enfoques entreguen el mismo resultado esperado. Las instrucciones `assert` detendrían la ejecución si alguna solución produjera un conteo diferente. También comprobaremos casos límite: una lista vacía y listas con una sola categoría de paridad.

````python
resultado_esperado = {"pares": 6, "impares": 6}

assert resultado_enfoque_1 == resultado_esperado
assert resultado_enfoque_2 == resultado_esperado
assert resultado_objetos == resultado_esperado
assert resultado_funcional == resultado_esperado

assert contar_paridad_funcional([]) == {"pares": 0, "impares": 0}
assert contar_paridad_funcional([2, 4, 6]) == {"pares": 3, "impares": 0}
assert contar_paridad_funcional([1, 3, 5]) == {"pares": 0, "impares": 3}

print("Todas las verificaciones se completaron correctamente.")
````

Los cuatro enfoques implementan la misma idea matemática: un entero es par si el residuo de su división entre $2$ es cero. Lo que cambia es la organización del programa. La versión imperativa hace explícito cada paso, la versión modular separa responsabilidades, la versión orientada a objetos asocia la operación con cada entero y la versión funcional expresa la transformación y el conteo mediante herramientas de orden superior.
